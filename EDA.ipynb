Section 5 – Missing value analysis

```python
missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print(missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

missing_table = missing_table[
    missing_table['Missing Values'] > 0
].sort_values(by='Missing Percentage', ascending=False)

display(missing_table)
```

```python
plt.figure(figsize=(10, 6))

missing_table['Missing Percentage'].plot(
    kind='bar',
    color='coral'
)

plt.title('Missing Values by Column')
plt.xlabel('Columns')
plt.ylabel('Missing Values (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Visualization 1 – Sales distribution

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['Total_Bill'],
    bins=30,
    color='skyblue',
    edgecolor='black'
)

plt.title('Distribution of Bakery Sales')
plt.xlabel('Total Bill Amount')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

Visualization 2 – Sales by product category

```python
category_sales = df.groupby('Category')['Total_Bill'].sum()

category_sales = category_sales.sort_values(
    ascending=False
)

plt.figure(figsize=(11, 6))

category_sales.plot(
    kind='bar',
    color='mediumseagreen'
)

plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Visualization 3 – Sales by hour

```python
hourly_sales = df.groupby('Hour')['Order_ID'].nunique()

plt.figure(figsize=(10, 6))

hourly_sales.plot(
    kind='line',
    marker='o',
    color='orange'
)

plt.title('Customer Orders by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Number of Unique Orders')

plt.xticks(range(0, 24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

Visualization 4 – Customer ratings

```python
rating_counts = df['Customer_Rating'].value_counts().sort_index()

plt.figure(figsize=(8, 5))

rating_counts.plot(
    kind='bar',
    color='mediumpurple'
)

plt.title('Distribution of Customer Ratings')
plt.xlabel('Customer Rating')
plt.ylabel('Number of Records')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Visualization 5 – Correlation heatmap

```python
plt.figure(figsize=(14, 10))

correlation = df.select_dtypes(
    include=['int64', 'float64']
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')

plt.tight_layout()
plt.show()
```

Visualization 6 – Waste percentage by category

```python
waste_category = df.groupby('Category')['Waste_Pct'].mean()

waste_category = waste_category.sort_values(
    ascending=False
)

plt.figure(figsize=(11, 6))

waste_category.plot(
    kind='bar',
    color='tomato'
)

plt.title('Average Waste Percentage by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Average Waste Percentage (%)')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
